```python
import numpy as np
import matplotlib.pyplot as plt
from yeti_iga.future import BSpline, BSplineSurface, BSplineVolume
```

Create a 1D BSpline parametric space of degree 2

```python
U = np.array([0.0, 0.0, 0.0, 1./3., 1./3., 2./3., 1., 1., 1.])
b = BSpline(2, U)

# Use getters
kv = ", ".join(f"{x:.3f}" for x in b.knot_vector)
print(f'Knot vector: [{kv}]')
print(f'Degree: {b.degree}')
```

Evaluate functions

```python
u = np.linspace(0., 1., 100)
n_basis = len(b.knot_vector) - b.degree - 1
colors = plt.cm.tab10(np.linspace(0, 1, n_basis))

fig, ax = plt.subplots()
for i in range(n_basis):
    N = [b.one_basis_fun(ui, i) for ui in u]
    ax.plot(u, N, color=colors[i], label=f'$N_{{{i}}}^{{{b.degree}}}$')

ax.set_xlabel('u')
ax.legend(loc='upper right')
plt.show()
```

Evaluate functions and derivatives for a given parameter value

```python
u = 0.5
span = b.find_span(u)

# Compute non-zero functions only
funs = b.basis_funs(span, u)
print(f'functions: {funs}')
print('-'*30)
# Compute non-zero functions and derivatives
dfuns = b.basis_funs_derivatives(span, u, 2)
print(f'functions: {dfuns[0, :]}')
print(f'1st derivatives: {dfuns[1, :]}')
print(f'2nd derivatives: {dfuns[2, :]}')

```

Create a 2D Bspline

```python
b1 = BSpline(2, np.array([0., 0., 0.,0.33, 0.66, 1., 1., 1.]))
b2 = BSpline(2, np.array([0., 0., 0., 0.2, 0.4, 0.6, 0.8, 1., 1., 1.]))

surf = BSplineSurface(b1, b2)

# Evaluate 2 parameters functions
u = np.array([0.3, 0.45])
span = surf.find_span_nd(u)
funs = surf.basis_funs_nd(span, u)
print(f'functions: {funs}')
```

Create a 3D BSpline

```python
b1 = BSpline(2, np.array([0., 0., 0.,0.33, 0.66, 1., 1., 1.]))
b2 = BSpline(2, np.array([0., 0., 0., 0.2, 0.4, 0.6, 0.8, 1., 1., 1.]))
# Different degree in 3rd direction
b3 = BSpline(3, np.array([0., 0., 0., 0., 0.3, 0.4, 0.8, 0.9, 1., 1., 1., 1.]))

vol = BSplineVolume(b1, b2, b3)

# Evaluate 3 parameters functions
u = np.array([0.3, 0.45, 0.72])
span = vol.find_span_nd(u)
funs = vol.basis_funs_nd(span, u)
# funs[i, j, k] = i-th function in direction 0 * j-th function in direction 1 * k-th function in direction 2
print(f'functions: {funs}')
```
